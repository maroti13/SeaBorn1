```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=sns.load_dataset("tips")
print(df.head())
```

```python
sns.lineplot(x="total_bill",y="tip",style="smoker", hue="sex",data=df)
plt.title("Total Bills Vs Tips")
plt.xlabel("Total Bill")
plt.ylabel("Tip")
plt.show()
```

```python
sns.lineplot(x="total_bill",y="tip",data=df)
plt.title("Total Bills Vs Tips")
plt.xlabel("Total Bill")
plt.ylabel("Tip")
plt.show()
```

```python
sns.lineplot(x="total_bill",y="tip",style="smoker",data=df)
plt.title("Total Bills Vs Tips")
plt.xlabel("Total Bill")
plt.ylabel("Tip")
plt.show()
```

```python
sns.lineplot(x="total_bill",y="tip",marker="o",data=df)
plt.title("Total Bills Vs Tips")
plt.xlabel("Total Bill")
plt.ylabel("Tip")
plt.show()
```

```python
average_tip=df.groupby("day")["tip"].mean() .reset_index()
sns.lineplot(x="day",y="tip",marker="o",data=average_tip)
plt.title("Total Bill Vs Average Tip")
plt.xlabel("total bill")
plt.ylabel("Average Tip")
plt.show()
```
